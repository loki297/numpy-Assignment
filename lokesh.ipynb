# Section A – Conceptual Understanding



### A1. Why are NumPy arrays preferred over Python lists?

NumPy arrays are mainly used for numerical calculations because they are faster than normal Python lists. They also take less memory and provide many built-in mathematical functions. Because of this, NumPy is very useful when working with large amounts of data in AI and ML.

### A2. Explain vectorization.

Vectorization means doing an operation on the complete NumPy array instead of using a `for` loop for each value. It makes the code shorter and usually much faster. This is especially useful when we have large datasets because NumPy can process many values at once. 


### A3. Difference between `A * B` and `A @ B`

`A * B` means multiplying the corresponding elements of two arrays. On the other hand, `A @ B` is used for matrix multiplication. In neural networks, `*` can be useful for element-wise operations, while `@` is commonly used to multiply inputs with weights.


### A4. What is broadcasting?

Broadcasting is a feature in NumPy that allows us to perform operations between arrays with different shapes. NumPy automatically adjusts the smaller array when their dimensions are compatible. Two dimensions are compatible when they are equal or when one of them is `1`.

### A5. Difference between `flatten()` and `ravel()`

Both `flatten()` and `ravel()` convert an array into a one-dimensional array. `flatten()` usually creates a new copy, while `ravel()` generally tries to use the existing data without making a copy. So, when memory is limited, I would prefer `ravel()` because it can use less extra memory.

### A7. Why do we normalize or standardize numerical features?

Normalization and standardization are used to bring numerical features to a similar scale.

| Normalization | Standardization |
|---|---|
| Converts values to a fixed range, usually 0 to 1. | Converts values based on the mean and standard deviation. |
| Uses the minimum and maximum values. | Uses the mean and standard deviation. |
| The values usually stay between 0 and 1. | Values can be negative or positive. |
| Useful when we need a fixed range of values. | Useful when the data has different distributions or scales. |
| Example result: `0.2, 0.5, 0.8` | Example result: `-1.2, 0, 1.3` |
| Commonly called Min-Max scaling. | Commonly called Z-score scaling. |

We use these techniques so that features with different ranges do not affect the ML model unfairly.

### A8. Why is `np.nanmean()` used?

`np.nanmean()` is useful when our data contains missing values represented by `NaN`. It calculates the mean while ignoring the `NaN` values. If we use `np.mean()` directly on data containing `NaN`, the result will normally be `NaN`.

#### D1. Data Preprocessing Pipeline

A data preprocessing pipeline is used to prepare raw data before it is given to a machine learning model.

### Step 1: Detect Missing Values

First, `np.isnan()` can be used to identify missing values in the NumPy array. It returns `True` for positions containing `NaN` and `False` for other values.

### Step 2: Handle Missing Values

After detecting the missing values, they need to be handled appropriately. In this example, the `NaN` values are replaced with the mean of the available values using `np.nanmean()`. This keeps all the data points instead of removing them.

### Step 3: Standardize the Clean Data

After handling the missing values, the clean data is standardized using its mean and standard deviation. Standardization changes the data so that it is centered around 0 and has a standard deviation of 1.

The complete preprocessing pipeline is:
1. Detect the missing values.
2. Calculate the mean while ignoring `NaN`.
3. Replace the `NaN` values with the mean.
4. Standardize the cleaned data.

`````python
import numpy as np

# Sample 1D array containing NaN values
data = np.array([10, 20, np.nan, 40, 50, np.nan])

# Step 1: Detect missing values
missing_values = np.isnan(data)

print("Missing values:")
print(missing_values)

# Step 2: Calculate mean ignoring NaN values
mean_value = np.nanmean(data)

# Replace NaN values with the mean
clean_data = np.where(np.isnan(data), mean_value, data)

print("\nMean used for replacement:", mean_value)
print("Clean data:")
print(clean_data)

# Step 3: Standardize the clean data
mean = np.mean(clean_data)
std = np.std(clean_data)

standardized_data = (clean_data - mean) / std

print("\nStandardized data:")
print(standardized_data)
`````

# Explain, with correct NumPy shapes, how a single RGB image differs from a batch of RGB images

#### A single RGB image is represented using three dimensions: **height, width, and color channels**. For example, if an RGB image is 64 × 64 pixels, its NumPy shape will be:

`(64, 64, 3)`

Here, the first `64` is the height, the second `64` is the width, and `3` represents the RGB channels.

When we have many RGB images, we store them together as a batch. In this case, one extra dimension is added at the beginning to represent the number of images. For example, a batch of 16 RGB images, each of size 64 × 64, will have the shape:

`(16, 64, 64, 3)`

Here, `16` is the number of images, `64` is the height, `64` is the width, and `3` represents the RGB channels.

### Shape Comparison

| Type | NumPy Shape | Meaning |
|---|---|---|
| Single RGB image | `(64, 64, 3)` | Height × Width × RGB channels |
| Batch of 16 RGB images | `(16, 64, 64, 3)` | Number of images × Height × Width × RGB channels |

So, the main difference is that a **single RGB image has 3 dimensions**, while a **batch of RGB images has 4 dimensions**. The extra first dimension tells us how many images are present in the batch.

`````python
import numpy as np

# Create a single RGB image of size 64 × 64
single_image = np.random.default_rng(42).integers(
    0, 256, size=(64, 64, 3)
)

# Create a batch of 16 RGB images
image_batch = np.random.default_rng(42).integers(
    0, 256, size=(16, 64, 64, 3)
)

print("Single RGB image shape:", single_image.shape)
print("Batch of RGB images shape:", image_batch.shape)
`````

# E1. Per-Column Min–Max Normalization 

### E1. Per-Column Min-Max Normalization

`````python
import numpy as np

X = np.array([[10, 100],
              [20, 200],
              [30, 300],
              [40, 400]], dtype=float)

min_val = X.min(axis=0)
max_val = X.max(axis=0)

normalized = (X - min_val) / (max_val - min_val)

print(normalized)
`````

### Explanation

First, we import NumPy using `import numpy as np`.

Then, we create a 2D array `X` with 4 rows and 2 columns. 
Here, each column represents a feature.

`X.min(axis=0)` finds the minimum value from each column.

`X.max(axis=0)` finds the maximum value from each column.

Next, we use the Min-Max Normalization formula:

**Normalized value = (X - minimum) / (maximum - minimum)**

The normalization is done column-wise because we use `axis=0`.

After normalization, all the values are converted into a range between `0` and `1`.

Finally, `print(normalized)` displays the normalized values.

The minimum value in each column becomes `0`, and the maximum value becomes `1`.
